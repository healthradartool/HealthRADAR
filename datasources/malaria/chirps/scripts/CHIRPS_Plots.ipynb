# CHIRPS Plots

This notebook produces the "What the Data Looks Like" plots for the CHIRPS data source page: the mean annual rainfall map for Southern Africa, and the monthly climatology and daily time series for Tete.

**Inputs (from `../data/`, produced by `CHIRPS_Download&Process.ipynb`):**
- `Fig1_PRCPTOT_daily_CHIRPS-3.0-0p05-rnl_199101-202012_SouthernAfrica.nc`
- `Fig2_PRCPTOT_mon_CHIRPS-3.0-0p05-rnl_199101-202012_climatology_Tete.csv`
- `Fig3_PRCPTOT_daily_CHIRPS-3.0-0p05-rnl_19810101-20260430_Tete.csv`

**Outputs:** `Fig1_CHIRPS_MeanAnnualRainfall_1991-2020.png`, `Fig2_CHIRPS_Climatology_Tete_1991-2020.png`, `Fig3_CHIRPS_TimeSeries_Rainfall_1981-2026.png`, saved to `../images/` - numbered to match their order on the data source page, mirroring the Vis1-Vis6 convention used in the Visualisations section.

## Step 1: Setup

```python
import os
import pandas as pd
import xarray as xr
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.pyplot as plt

# Southern Africa map extent, used to frame the first plot
lat_min, lat_max = -35, 0
lon_min, lon_max = 10, 42

# Make sure the output folder exists before we try to save anything into it
os.makedirs('../images', exist_ok=True)
```

## Mean Annual Rainfall - Southern Africa

Spatial map of mean annual rainfall across Southern Africa, averaged over the 1991-2020 climatology period.

```python
mean_prcptot = xr.open_dataarray(
    '../data/Fig1_PRCPTOT_daily_CHIRPS-3.0-0p05-rnl_199101-202012_SouthernAfrica.nc'
)
```

```python
plt.figure(figsize=(10, 6))
ax = plt.axes(projection=ccrs.PlateCarree())

pcm = ax.pcolormesh(
    mean_prcptot['lon'], mean_prcptot['lat'], mean_prcptot,
    cmap='YlGnBu', shading='auto', transform=ccrs.PlateCarree()
)

ax.add_feature(cfeature.COASTLINE, edgecolor='black', linewidth=1)
ax.add_feature(cfeature.BORDERS, edgecolor='black', linewidth=1)

plt.colorbar(pcm, ax=ax, label='Mean Annual Rainfall (mm/year)')

gl = ax.gridlines(
    draw_labels=True, crs=ccrs.PlateCarree(),
    linewidth=0.5, color='gray', alpha=0.7, linestyle='--'
)
gl.top_labels = False
gl.right_labels = False
gl.xlabel_style = {'size': 10}
gl.ylabel_style = {'size': 10}

ax.set_title('Mean Annual Rainfall for Southern Africa (1991-2020) - CHIRPS v3')
ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

plt.savefig('../images/Fig1_CHIRPS_MeanAnnualRainfall_1991-2020.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Monthly Climatology - Tete

Average rainfall for each calendar month at Tete, 1991-2020.

```python
rainfall_climatology = pd.read_csv(
    '../data/Fig2_PRCPTOT_mon_CHIRPS-3.0-0p05-rnl_199101-202012_climatology_Tete.csv',
    index_col=0
)
```

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

color = 'tab:blue'
ax1.set_xlabel('Month')
ax1.set_ylabel('Rainfall (mm)')
ax1.bar(
    rainfall_climatology.index, rainfall_climatology['prcptot'],
    width=0.4, color=color, label='Rainfall (mm)', zorder=3
)
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(axis='y', linewidth=0.7, zorder=0)

ax1.set_xticks(range(1, 13))
ax1.set_xticklabels([
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
])

fig.suptitle('Monthly Rainfall Climatology for Tete, MOZ (1991-2020) - CHIRPS v3')

plt.savefig('../images/Fig2_CHIRPS_Climatology_Tete_1991-2020.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Daily Time Series - Tete

Full daily rainfall record for Tete, 1981-2026.

```python
rainfall_daily_df = pd.read_csv(
    '../data/Fig3_PRCPTOT_daily_CHIRPS-3.0-0p05-rnl_19810101-20260430_Tete.csv',
    index_col=0,
    parse_dates=True
)
```

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

color = 'tab:blue'
ax1.set_xlabel('Time')
ax1.set_ylabel('Rainfall (mm)')
ax1.bar(
    rainfall_daily_df.index, rainfall_daily_df['prcptot'],
    color=color, label='Rainfall (mm)', width=15
)
ax1.tick_params(axis='y', labelcolor=color)
ax1.legend(loc='upper left')

fig.suptitle('Daily Rainfall for Tete, MOZ (1981-2026) - CHIRPS v3')
ax1.grid(True)

plt.savefig('../images/Fig3_CHIRPS_TimeSeries_Rainfall_1981-2026.png', dpi=300, bbox_inches='tight')
plt.show()
```
